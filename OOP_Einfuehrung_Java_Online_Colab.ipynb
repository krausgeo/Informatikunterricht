## Kapitel 1 — Datentypen und Variablen

Beispiele werden als Java-Dateien kompiliert und ausgeführt.

```python
# Beispiel: Datentypen in einer Java-Datei
java_code = """public class TypesDemo {
    public static void main(String[] args) {
        int a = 5;
        double b = 3.14;
        boolean ok = true;
        char c = 'A';
        String s = "Hallo";
        System.out.println("a=" + a);
        System.out.println("b=" + b);
        System.out.println("ok=" + ok);
        System.out.println("c=" + c);
        System.out.println("s=" + s);
    }
}
"""
with open("TypesDemo.java","w", encoding="utf-8") as f:
    f.write(java_code)
!javac TypesDemo.java
!java TypesDemo
```

## Kapitel 2 — Klassen und Objekte

Beispiel `Person`: Konstruktor, Getter/Setter, Methode.

```python
person_code = """class PersonDemo {
    public static void main(String[] args) {
        Person p = new Person("Anna", 17);
        p.greet();
        p.setAge(18);
        System.out.println("Nach Geburtstag: " + p.getAge());
    }
}

class Person {
    String name;
    int age;
    Person(String name, int age) { this.name = name; this.age = age; }
    String getName() { return name; }
    int getAge() { return age; }
    void setName(String name) { this.name = name; }
    void setAge(int age) { this.age = age; }
    void greet() {
        System.out.println("Hallo, ich heiße " + name + " und bin " + age + " Jahre alt.");
    }
}
"""
with open("PersonDemo.java","w", encoding="utf-8") as f:
    f.write(person_code)
!javac PersonDemo.java
!java PersonDemo
```

## Kapitel 3 — Methoden und Rückgabewerte

Beispiel `MathUtil.add` als Java-Klasse.

```python
math_code = """class MathUtil {
    public static int add(int x, int y) {
        int result = x + y;
        return result;
    }
}
class MathDemo {
    public static void main(String[] args) {
        int sum = MathUtil.add(4,7);
        System.out.println("Summe: " + sum);
    }
}
"""
with open("MathDemo.java","w", encoding="utf-8") as f:
    f.write(math_code)
!javac MathDemo.java
!java MathDemo
```

## Kapitel 4 — Algorithmus & Pseudocode

Pseudocode-Beispiel und Java-Implementierung (if/else).

```python
algo_code = """public class AlgoDemo {
    public static void main(String[] args) {
        int x = -3;
        if (x > 0) {
            System.out.println("positiv");
        } else {
            System.out.println("nicht positiv");
        }
    }
}
"""
with open("AlgoDemo.java","w", encoding="utf-8") as f:
    f.write(algo_code)
!javac AlgoDemo.java
!java AlgoDemo
```

## Kapitel 5 — Vererbung

Beispiel mit `Fahrzeug` und `Auto`.

```python
inherit_code = """class Fahrzeug {
    String marke;
    Fahrzeug(String marke) { this.marke = marke; }
    void fahren() { System.out.println("Das Fahrzeug fährt."); }
}
class Auto extends Fahrzeug {
    int tueren;
    Auto(String marke, int tueren) { super(marke); this.tueren = tueren; }
    @Override
    void fahren() { System.out.println("Das Auto " + marke + " fährt mit " + tueren + " Türen."); }
}
class InheritDemo {
    public static void main(String[] args) {
        Fahrzeug f = new Fahrzeug("Allgemein");
        f.fahren();
        Auto a = new Auto("VW", 4);
        a.fahren();
    }
}
"""
with open("InheritDemo.java","w", encoding="utf-8") as f:
    f.write(inherit_code)
!javac InheritDemo.java
!java InheritDemo
```

## Dynamische Visualisierung mit Graphviz

Beispiel: Klassendiagramm (wird mit dem Python `graphviz`-Paket gerendert).

```python
from graphviz import Digraph
g = Digraph('G', filename='klassendiagramm', format='svg')
g.attr('node', shape='record')
g.node('Fahrzeug', '<f0> Fahrzeug | <f1> - marke: String \l - speed: int \l | <f2> + fahren()')
g.node('Auto', '<f0> Auto | <f1> - tueren: int \l | <f2> + fahren()')
g.edge('Fahrzeug', 'Auto', arrowhead='empty', label='Vererbung')
g
```
