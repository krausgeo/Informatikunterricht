## Kapitel 2 — Klassen und Objekte

```java
class Person {
    String name;
    int age;
    Person(String name, int age) { this.name = name; this.age = age; }
    String getName() { return name; }
    int getAge() { return age; }
    void setName(String name) { this.name = name; }
    void setAge(int age) { this.age = age; }
    void greet() { System.out.println("Hallo, ich heiße " + name + " und bin " + age + " Jahre alt."); }
}
Person p = new Person("Anna", 17);
p.greet();
```

## Kapitel 3 — Methoden: Kopf, Rumpf, Aufruf, Rückgabewert

```java
class MathUtil {
    public static int add(int x, int y) {
        int result = x + y;
        return result;
    }
}
int sum = MathUtil.add(4, 7);
System.out.println("Summe: " + sum);
```

## Kapitel 5 — Vererbung, Generalisierung & Spezialisierung

```java
class Fahrzeug {
    String marke;
    Fahrzeug(String marke) { this.marke = marke; }
    void fahren() { System.out.println("Das Fahrzeug fährt."); }
}
class Auto extends Fahrzeug {
    int tueren;
    Auto(String marke, int tueren) { super(marke); this.tueren = tueren; }
    @Override
    void fahren() { System.out.println("Das Auto " + marke + " fährt mit " + tueren + " Türen."); }
}
Fahrzeug f = new Fahrzeug("Allgemein"); f.fahren();
Auto a = new Auto("VW", 4); a.fahren();
```

## Kapitel 6 — Dynamische Visualisierung mit Graphviz

```java
from graphviz import Digraph

g = Digraph('G', format='svg')
g.attr('node', shape='record')
g.node('Fahrzeug', '<f0> Fahrzeug | <f1> - marke: String \l - speed: int \l | <f2> + fahren()')
g.node('Auto', '<f0> Auto | <f1> - tueren: int \l | <f2> + fahren()')
g.edge('Fahrzeug', 'Auto', arrowhead='empty', label='Vererbung')
g
```
