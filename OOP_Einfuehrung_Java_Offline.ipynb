## Kapitel 2 — Klassen und Objekte

Beispielklasse `Person` mit Attributen, Konstruktor, Getter/Setter und Methode `greet()`.

````java
class Person {
    String name;
    int age;

    // Konstruktor
    Person(String name, int age) {
        this.name = name;
        this.age = age;
    }

    // Getter
    String getName() { return name; }
    int getAge() { return age; }

    // Setter
    void setName(String name) { this.name = name; }
    void setAge(int age) { this.age = age; }

    // Methode
    void greet() {
        System.out.println("Hallo, ich heiße " + name + " und bin " + age + " Jahre alt.");
    }
}

// Objekt erstellen und benutzen
Person p = new Person("Anna", 17);
p.greet();
````

## Kapitel 3 — Methoden: Kopf, Rumpf, Aufruf, Rückgabewert

Beispiel einer Methode mit Rückgabewert:

````java
class MathUtil {
    // Methodenkopf: public static int add(int x, int y)
    // Methodenrumpf: Körper zwischen { ... }
    public static int add(int x, int y) {
        int result = x + y; // lokale Variable
        return result; // Rückgabewert
    }
}

int sum = MathUtil.add(4, 7);
System.out.println("Summe: " + sum);
````

## Kapitel 5 — Vererbung, Generalisierung & Spezialisierung

Beispiel: Oberklasse `Fahrzeug`, Unterklasse `Auto` und Überschreiben einer Methode.

````java
class Fahrzeug {
    String marke;
    Fahrzeug(String marke) { this.marke = marke; }
    void fahren() { System.out.println("Das Fahrzeug fährt."); }
}

class Auto extends Fahrzeug {
    int tueren;
    Auto(String marke, int tueren) {
        super(marke);
        this.tueren = tueren;
    }
    @Override
    void fahren() { System.out.println("Das Auto " + marke + " fährt mit " + tueren + " Türen."); }
}

Fahrzeug f = new Fahrzeug("Allgemein");
f.fahren();
Auto a = new Auto("VW", 4);
a.fahren();
````
